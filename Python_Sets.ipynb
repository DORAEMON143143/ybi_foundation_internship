# Python Sets

```python
my_set = {1, 2, 3, 4, 5}
print(f"Original set: {my_set}")

# Adding elements
my_set.add(6)
print(f"Set after adding 6: {my_set}")

# Adding an existing element (no change)
my_set.add(3)
print(f"Set after adding 3 again: {my_set}")

# Removing elements
my_set.remove(2)
print(f"Set after removing 2: {my_set}")

# Union of two sets
other_set = {4, 5, 6, 7, 8}
union_set = my_set.union(other_set)
print(f"Union with {{4, 5, 6, 7, 8}}: {union_set}")

# Intersection of two sets
intersection_set = my_set.intersection(other_set)
print(f"Intersection with {{4, 5, 6, 7, 8}}: {intersection_set}")

# Difference of two sets
difference_set = my_set.difference(other_set)
print(f"Difference with {{4, 5, 6, 7, 8}}: {difference_set}")
```

```python
# Checking for element existence
print(f"Is 5 in my_set? {5 in my_set}")
print(f"Is 10 in my_set? {10 in my_set}")

# Using the 'not in' operator
print(f"Is 7 not in my_set? {7 not in my_set}")
```

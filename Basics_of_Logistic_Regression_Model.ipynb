# Basics of Logistic Regression Model

### What is Logistic Regression?

**Logistic Regression** is a fundamental statistical and machine learning algorithm used for **binary classification** tasks (predicting one of two possible outcomes, like Yes/No, Spam/Not Spam, or Success/Failure).

Unlike Linear Regression, which predicts continuous values (e.g., house prices), Logistic Regression predicts the **probability** ($p$) that a given input belongs to a specific class (usually labeled as `1`).

---

### Key Concepts

#### 1. The Logistic (Sigmoid) Function
To map any real-valued number to a probability between $0$ and $1$, logistic regression uses the **Sigmoid function**:

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

* If $z$ is a large positive number, $\sigma(z)$ approaches $1$.
* If $z$ is a large negative number, $\sigma(z)$ approaches $0$.
* If $z = 0$, $\sigma(z) = 0.5$.

#### 2. The Model Equation
In logistic regression, we model the input $z$ as a linear combination of the input features ($x_1, x_2, ...$):

$$z = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \dots + \beta_n x_n$$

Substituting $z$ into the sigmoid function gives the predicted probability:

$$P(Y=1|X) = \frac{1}{1 + e^{-(\beta_0 + \beta_1 x_1 + \dots + \beta_n x_n)}}$$

#### 3. Decision Boundary
To make a final classification, we set a **threshold** (usually $0.5$):
* If $P(Y=1|X) \ge 0.5$, we predict class **1**.
* If $P(Y=1|X) < 0.5$, we predict class **0**.

#### 4. Odds and Log-Odds (Logit)
* **Odds**: The ratio of the probability of success to the probability of failure:
  $$\text{Odds} = \frac{p}{1-p}$$
* **Log-Odds (Logit)**: The natural logarithm of the odds, which is linear with respect to the features:
  $$\ln\left(\frac{p}{1-p}\right) = \beta_0 + \beta_1 x_1 + \dots + \beta_n x_n$$

#### 5. Loss Function: Binary Cross-Entropy
Logistic regression is trained using **Maximum Likelihood Estimation (MLE)** to find the parameters $\beta$ that minimize the **Log Loss** (Binary Cross-Entropy):

$$\text{Loss} = -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} \ln(p^{(i)}) + (1 - y^{(i)}) \ln(1 - p^{(i)}) \right]$$

This loss penalizes highly confident incorrect predictions.

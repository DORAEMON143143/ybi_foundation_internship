# Python Dictionary

```python
# Python Dictionary Examples

# 1. Creating a dictionary
person = {
    "name": "Alice",
    "age": 30,
    "city": "New York",
    "skills": ["Python", "Data Science"]
}
print("Original dictionary:", person)

# 2. Accessing values
print("Name:", person["name"])
print("Age (using get()):", person.get("age"))
# get() returns None (or a default value) instead of raising a KeyError if the key doesn't exist
print("Country (with default):", person.get("country", "USA"))

# 3. Modifying and adding items
person["age"] = 31  # Update existing key
person["occupation"] = "Engineer"  # Add new key-value pair
print("\nModified dictionary:", person)

# 4. Removing items
removed_value = person.pop("city")  # Removes 'city' and returns its value
print(f"\nRemoved 'city' (value: {removed_value})")
print("Dictionary after pop:", person)

# 5. Iterating through a dictionary
print("\nIterating through keys and values:")
for key, value in person.items():
    print(f"- {key}: {value}")
```

# Python Lists

```python
listexample = [1, 2, 3, "apple", "banana"]
print(listexample)
print(f"Length of the list: {len(listexample)}")
print(f"First element: {listexample[0]}")
print(f"Last element: {listexample[-1]}")
```

### Appending Items to a List

The `append()` method adds a single item to the end of an existing list.

```python
print(f"Original list: {listexample}")

# Append a single item
listexample.append("orange")
print(f"List after appending 'orange': {listexample}")

# Append another item
listexample.append(4)
print(f"List after appending 4: {listexample}")
```
